**The *else* statement**

**The *elif* statement** (for multiple conditions)

```python
# Basic syntax

# if condition1:
    # code if condition1 is True
# elif condition2:
    # code if condition2 is True
# else:
    # code if neither condition is True
    
```

```python
# Example of if-elif-else statement
# Before running: What will happen and why?

# Answer: the elif statement will activate since x = 5 is true

x = 5
if x > 5:
    print("x is greater than 5")
elif x == 5:
    print("x is equal to 5")
else:
    print("x is less than 5")
```

## Loops
Loops in Python are used to execute a block of code repeatedly. We will look at two different types of loops; the *for loop*, and the *while loop*.  
**The for loop** iterates over sequences, so it keeps executing until it has reached the end of the loop.  
**The while loop** keeps repeating while a condition is true, and first ends when the condition becomes false. 

```python
# For loop - Basic syntax
# for element in sequence:
    # code to execute for each element

# Example of for loop
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(len(fruit))
```

```python
# While loop - Basic syntax
# while condition:
    # code to execute while condition is True
    
# Example of while loop
x = 1
while x < 6:
    print(x)
    x += 1
```

Now lets add some extra requirements:
- The password needs to have at least 3 digits
- And there needs to be an uppercase letter

```python
# Modified code here

def password_checker():
    password = input("Write password")
    digits = 0
    uppercase = 0

    for n in password:
        if n.isdigit():
            digits += 1
        elif n.isupper():
            uppercase += 1
        
    
    print(digits)
    print(uppercase)

    if len(password) < 8 or len(password) > 32:
        print("wrongggg")
    elif digits < 3:
        print("Error: The password needs at least 3 digits")
    elif uppercase == 0:
        print("Error: There needs to be at least one uppercase letter")
    else:
        print("this is greatt")

password_checker()
```

And the last bits, now we also need to ensure that the password contains:
- 2 special characters

Make sure the user understands why a wrong password is rejected. And ensure that only a password that holds all requirements are passed. 

```python
# Last modifications on the code - final password_checker

def password_checker():
    password = input("Write password")
    digits = 0
    uppercase = 0
    specials = 0

    for n in password:
        if n.isdigit():
            digits += 1
        elif n.isupper():
            uppercase += 1
        elif n == "%" or n == "#" or n == "@":
            specials += 1

    
    print(password)
    print(digits)
    print(uppercase)
    print(specials)

    if len(password) < 8 or len(password) > 32:
        print("wrongggg")
    elif digits < 3:
        print("Error: The password needs at least 3 digits")
    elif uppercase == 0:
        print("Error: There needs to be at least one uppercase letter")
    elif specials < 2:
        print("Error: You need at least 2 special characters (%, #, or @)")
    else:
        print("this is greatt")

password_checker()

```

Now make a function that does the opposite.   
Hint: C = ((fahrenheit-32)/(9/5))  

```python
# Fahrenheit to Celsius converter

def convert_fahrenheit(f):
    c = ((f-32)/(9/5))
    print(c)

convert_fahrenheit(100)

```

Well done! 
Now wouldn't it be neat if we could just wrap that into a function that we could call temperature_converter with two parameters "temperature" and "unit" so that we could convert both ways. 



```python
# Temperature converter here

def temp_converter(temperature, unit):
    temp = temperature
    unit = unit

    if unit == "c" or unit == "celsius": # if the unit is celsius we want it converted to fahrenheit
        temp = ((temp*9/5)+32)
        print(temp)

    elif unit == "f" or unit == "fahrenheit": # if the unit is fahrenheit we want it converted to celsius
        temp = ((temp-32)/(9/5))
        print(temp)
        
    else: # error messeage for if the unit is not stated correctly
        print("error")

temp_converter(100, "fahrenheit") # Test: Here we want to convert 100 fahrenheit to celsius. The output should be 37.8
    

```

## Grade calculator

The scenario: You’re helping a teacher who wants to calculate student grades quickly. The teacher has been doing these calculations by hand, which takes a long time and sometimes leads to mistakes.

The challenge: 
- Calculate the average of test scores
- Convert percentage grades to letter grades (it's just easier than the danish grading metric ngl)
- Calculate final grades with different weights for different assignments

Example data:  
Tests count for 70%, homework counts for 30%  

Test scores: 85, 93, 78, 96, 88
Homework average: 90

Now we start with one thing first. Each functions should do _one_ thing effectively. 


```python
# Make a function that calculate the average of test scores. 

def calculate_average(testscores):
    totalscores = sum(testscores)
    average = totalscores / len(testscores)
    return average

calculate_average([85, 93, 78, 96, 88]) # Test: Here we want to calculate the average of the test scores.
```

Well done, now we move on to the next sub-task

Hint: Grade to lettter grade converter  
90–100 → A  
80–89  → B  
70–79  → C  
60–69  → D  
0–59   → F  


```python
# Make a function that converst the test scores to letter grades

def percentage_to_letter_grade(average_score):
    if average_score >= 90:
        return "A" 
    elif average_score >= 80:
        return "B"
    elif average_score >= 70:
        return "C"
    elif average_score >= 60:
        return "D"
    else:
        return "F"

    
    
# Test with different percentages
percentage_to_letter_grade(95) # Test: Here we want to convert 95% to a letter grade. The output should be "A"
percentage_to_letter_grade(85) # Test: Here we want to convert 85% to a letter grade. The output should be "B"
percentage_to_letter_grade(75) # Test: Here we want to convert 75% to a letter grade. The output should be "C"
```

Last task, We need to calculate the weighted grade


```python
def calculate_weighted_grade(testscores, homeworkscores):
    weighted_score = (testscores*0.7) + (homeworkscores*0.3)  
    return weighted_score

calculate_weighted_grade(85, 90) # Test: Here we want to calculate the weighted grade with 85% test score and 90% homework score
```

Now we want to give the professor one function that uses all these three and wraps them in a neat little function with everything put together. 

```python
def calculate_student_grade(testscores, homeworkscores):

    # What is the average of the test scores of the entire class?
    average_test_score = sum(testscores) / len(testscores)
    print(f"The average test score of the class is: {average_test_score}")
    
    # What are the letter grades of the students based on their test scores?
    lettergrades = []
    for score in testscores:
        if score >= 90:
            lettergrades.append("A")
        elif score >= 80:
            lettergrades.append("B")
        elif score >= 70:
            lettergrades.append("C")
        elif score >= 60:
            lettergrades.append("D")
        else:
            lettergrades.append("F")
    print(f"The letter grades of the students are: {lettergrades}")

    # What are the weighted grades of the students based on their test scores and homework scores?
    weighted_scores = []
    for score in testscores:
        weighted_score = (score*0.7) + (homeworkscores*0.3)
        weighted_scores.append(weighted_score)
    print(f"The weighted grades of the students are: {weighted_scores}")

calculate_student_grade([85, 93, 78, 96, 88], 90)    
```
